Stack(Linked-List)

```python
class Node:
  def __init__(self, data): #Node객체를 초기화하는 생성자
    self.data = data
    self.next = None #next는 다음 노드를 가리키는 포인터
```

```python
class LinkedListStack:
  def __init__(self):
    self.top = None

  def push(self, data): #맨 위에 노드 추가
    new_node = Node(data) #스택에 추가할 데이터를 포함하는 새 Node객체 생성
    new_node.next = self.top #새 노드의 next포인터가 현재 스택의 top을 가리키도록 설정
    #첫 노드->None(현재 top) OR 새 노드->이전 노드(현재top)
    self.top = new_node #첫 노드(top), 새 노드(top)->이전 노드
    print(f"Pushed: {data}")

  def is_empty(self): #현재 스택이 비어 있는지 확인
    return self.top is None #결과에 따라 T/F반환

  def pop(self): #데이터 제거와 반환
    if self.is_empty():
      print("Stack is empty. Cannot pop.")
    else:
      pop_data = self.top.data # 현재 top의 데이터를 저장
      self.top = self.top.next # top을 다음 노드로 이동시켜 현재 노드(연결고리에서 빠짐)를 제거
      print(f"Popped: {pop_data}") #return뒤 코드 실행x
      return pop_data #데이터 반환

  def display(self):
    print("Stack (top to bottom):")
    if self.is_empty():
      print("None")
    else:
      display_node = self.top
      while display_node is not None:
        print(f"  -> {display_node.data}")
        display_node = display_node.next

  def peek(self): #가장 위 데이터 확인
    if self.is_empty():
      print("Stack is empty.")
    else:
      print(f"Peeked: {self.top.data}")
      return self.top.data
```

```python
#스택 객체 생성
stack = LinkedListStack()

#데이터 push
stack.push(10)
stack.push(20)
stack.push(30)

print("-" * 20)

#스택 상태 확인
stack.display()

print("-" * 20)

#가장 위 데이터 확인
stack.peek()

print("-" * 20)

#데이터 pop
stack.pop()
stack.pop()

print("-" * 20)

#다시 스택 상태 확인
stack.display()

print("-" * 20)

#비어있는 스택에서 pop시도
stack.pop()
stack.pop() #한 번 더 시도
```

Stack(Fixed Array)

```python
class FixedArrayStack:
  def __init__(self, capacity=5):
    self.capacity = capacity
    self.items = [None] * capacity #데이터를 저장할 리스트 [None,None,...]
    self.top = -1

  def is_empty(self):
    return self.top == -1

  def is_full(self):
    return self.top == self.capacity-1

  def push(self, data): #none을 데이터로 교체
    if self.is_full():
      print("Stack Overflow! Cannot push to a full stack.")
    else:
      self.top += 1
      self.items[self.top] = data
      print(f"Pushed: {data}")

  def pop(self):
    if self.is_empty():
      print("Stack Underflow! Cannot pop from an empty stack.")
    else:
      pop_data = self.items[self.top]
      print(f"Popped: {pop_data}")
      self.items[self.top] = None
      self.top -= 1
      return pop_data

  def peek(self):
    if self.is_empty():
      print("Stack is empty.")
    else:
      print(f"Peeked: {self.items[self.top]}")
      return self.items[self.top]

  def display(self):
    print(f"Stack (size={self.top+1}/capacity={self.capacity}):")
    if self.is_empty():
      print("None")
    else:
      position = self.top
      while position > -1:
        print(f"  -> {self.items[position]}")
        position -= 1
```

```python
#용량이 3인 스택 객체 생성
stack = FixedArrayStack(3)

#데이터 push (성공)
stack.push(10)
stack.push(20)
stack.push(30)

print("-" * 20)

#스택 상태 확인 (가득 참)
stack.display()
print(f"Is full? {stack.is_full()}")

print("-" * 20)

#스택 오버프로우 테스트
stack.push(40) #용량을 초과하여 push 시도

print("-" * 20)

#데이터 pop
stack.pop()
stack.peek()
stack.pop()

print("-" * 20)

#다시 스택 상태 확인
stack.display()

print("-" * 20)

#스택 언더플로우 테스트
stack.pop() #마지막 데이터 pop
stack.pop() #비어있는 상태에서 pop 시도
```

Stack(Resizing Array)

```python
class DynamicArrayStack:
  def __init__(self, capacity=5):
    self.capacity = capacity
    self.items = [None] * capacity
    self.top = -1

  def is_empty(self):
    return self.top == -1

  def is_full(self):
    return self.top == self.capacity-1

  def resize(self):
    new_capacity = self.capacity * 2
    new_items = [None] * new_capacity

    for i in range(0,self.capacity): #기존 배열 복사
      new_items[i] = self.items[i]

    self.capacity = new_capacity #기존 객체의 속성값을 새로운 값으로 교체
    self.items = new_items

  def push(self, data): #none을 데이터로 교체
    if self.is_full():
      self.resize()

    self.top += 1
    self.items[self.top] = data
    print(f"Pushed: {data}")

  def pop(self):
    if self.is_empty():
      print("Stack Underflow! Cannot pop from an empty stack.")
    else:
      pop_data = self.items[self.top]
      print(f"Popped: {pop_data}")
      self.items[self.top] = None
      self.top -= 1
      return pop_data

  def peek(self):
    if self.is_empty():
      print("Stack is empty.")
    else:
      print(f"Peeked: {self.items[self.top]}")
      return self.items[self.top]

  def display(self):
    print(f"Stack (size={self.top+1}/capacity={self.capacity}):")
    if self.is_empty():
      print("None")
    else:
      position = self.top
      while position > -1:
        print(f"  -> {self.items[position]}")
        position -= 1
```

```python
#용량이 3인 스택 객체 생성
stack = DynamicArrayStack(3)

#데이터 push (성공)
stack.push(10)
stack.push(20)
stack.push(30)

print("-" * 20)

#스택 상태 확인 (가득 참)
stack.display()
print(f"Is full? {stack.is_full()}")

print("-" * 20)

#스택 resize 테스트
stack.push(40) #용량을 초과하여 push 시도

print("-" * 20)

#데이터 pop
stack.pop()
stack.pop()
stack.peek()
stack.pop()

print("-" * 20)

#다시 스택 상태 확인
stack.display()

print("-" * 20)

#스택 언더플로우 테스트
stack.pop() #마지막 데이터 pop
stack.pop() #비어있는 상태에서 pop 시도
```

CheckBrackets

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class CheckBrackets_Stack:
  def __init__(self):
    self.top = None

  def push(self, data):
    new_node = Node(data)
    new_node.next = self.top
    self.top = new_node

  def is_empty(self):
    return self.top is None

  def pop(self):
    if self.is_empty():
      return False #닫는 괄호가 나왔는데 Stack이 비어 있음
    else:
      pop_data = self.top.data
      self.top = self.top.next
      return pop_data # open_brackets[i]와 같은지 확인
```

```python
"""LinkedListStack()을 이용한 CheckBrackets_Stack 사용
문자열 반복문
여는 괄호(,{,[ -> push
닫는 괄호],}.) -> pop해서 짝 확인
if 짝x -> False
모든 문자를 확인한 후 스택이 비어 있으면 True"""

def CheckBrackets(case):
    open_brackets = ["(", "{", "["]
    close_brackets = [")", "}", "]"]
    stack = CheckBrackets_Stack()

    for letter in case:
        # 여는 괄호(,{,[ -> push
        if letter in open_brackets:
            stack.push(letter)

        # 닫는 괄호],}.) -> pop해서 짝 확인
        elif letter in close_brackets:
            i = close_brackets.index(letter)
            # 스택이 비어있는지 확인하거나 꺼낸 괄호와 짝이 맞는지 검사
            if stack.pop() != open_brackets[i]:
                return False #return문 실행시 함수 종료-> return stack.is_empty()실행x

    # 모든 문자를 확인한 후 스택이 비어 있으면 True
    return stack.is_empty()
```

```python
test_cases = [
"()", # True
"()[]{}", # True
"({[]})", # True
"([)]", # False (짝이 어긋남)
"((())", # False (여는 괄호가 더 많음)
"())", # False (닫는 괄호가 더 많음)
"{[}", # False
"print('Hello, World!')", # True (문자열 내의 다른 문자는 무시)
"if (x > 5) and (y < 3):", # True
"", # True (빈 문자열)
]
for case in test_cases:
  print(f"{case}: {CheckBrackets(case)}")
```
