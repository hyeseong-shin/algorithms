Singly linked list

```python
class Node:
  """A node holding one element"""
  def __init__(self,data):
    self.data = data
    self.next = None
```

```python
class SinglyLinkedList:
  """Singly linked list implementation"""
  def __init__(self):
    self.head = None

  def is_empty(self):
    return self.head is None

  def insert(self, pos: int, item: int):
    if pos < 0:
      return False

    new_node = Node(item)

    #맨 앞에 삽입하는 경우 ex)a-b insert(0,c)
    if pos == 0:
      new_node.next = self.head #맨 앞에 노드가 있으면 삽입 노드 다음으로 ex)c.next=a
      self.head = new_node
      #->결과:c-a-b

    #중간에 삽입하는 경우 ex)a-b-c insert(2,f)
    else:
      if self.is_empty():
        return False

      current = self.head #포인터 이동 위치
      for i in range(pos - 1): #삽입위치 전까지 이동 ex)current=b
        current = current.next #for문이 돌 때마다 한 칸씩 이동
        if current is None:#list는 full
          return False

      #current의 포인터(원래 노드에 연결)를 새 노드의 포인터로 대체
      #->원래 노드가 새 노드의 다음에 ex)f.next=c
      new_node.next = current.next
      #current의 포인터를 새 노드로 연결  ex)b.next=f
      current.next = new_node
      #ex)결과: a-b-f-c

  def delete(self, pos: int):
    if pos < 0:
      return False

  #삭제 노드 이전 노드의 포인터를 삭제 노드에서 그 다음 노드로 연결
    if self.is_empty():
        return False

    if pos == 0:
      delete_node = self.head
      self.head = self.head.next

    else:
      current = self.head
      for i in range(pos-1):
        current = current.next
        if current is None:
          return False

      if current.next is None:
        return False
      delete_node = current.next
      current.next = current.next.next

    return delete_node.data

  def get_entry(self, pos: int):
    if pos < 0:
      return None

    if self.is_empty():
      return None

    current = self.head
    for i in range(pos):
      current = current.next
      if current is None:
        return None

    return current.data #data가 반환하고자 하는 값

  def print_forward(self):
    if self.is_empty():
      print("None")
    else:
      current = self.head
      while current is not None:
        print(current.data, end=" ")
        current = current.next
      print()
```

```python
sll = SinglyLinkedList()
print("is_empty:", sll. is_empty()) # True

sll.insert (0, 10)
sll.insert (1, 20)
sll.insert (1, 15)

print("List forward:", end=" ")
sll.print_forward()

print("get_entry(0):", sll.get_entry(0)) # 10
print("get_entry(1):", sll.get_entry(1))#15
print("get_entry(2):", sll.get_entry(2))# 20
print("get_entry(3):", sll.get_entry(3))# None



print("delete at 1:", sll.delete(1))# 15
print("List after delete:", end=" ")
sll.print_forward()

print("get_entry (1):", sll.get_entry(1))# 20
print ("get_entry(2):", sll.get_entry(2))# None

print("delete at 0:", sll.delete(0))# 10
print("delete at 0:", sll.delete(0))# 20
print("is_empty:", sll.is_empty()) # True
```

Binary Tree

```python
class BinaryTreeNode:
  """A node of a binary tree"""
  def __init__(self,data):
    self.data = data
    self.left = None
    self.right = None

  def add_left(self, Node): #n2가 Node
    self.left = Node #ex)root.add_left(n2)에서 root가 self

  def add_right(self, Node):
    self.right = Node

  def preorder_traversal(self): #root-left-right
    print(self.data, end=" ") #현재 노드

    if self.left is not None:
      self.left.preorder_traversal() #같은 함수를 다시 호출해 반복적 순회

    if self.right is not None:
      self.right.preorder_traversal()
```

```python
print("\nBinary tree structure:")
root = BinaryTreeNode('1')
n2 = BinaryTreeNode ('2')
n3 = BinaryTreeNode ('3')
n4 = BinaryTreeNode ('4')
n5 = BinaryTreeNode('5')
n6 = BinaryTreeNode('6')
n7 = BinaryTreeNode('7')

root.add_left(n2)
root.add_right(n3)
n2.add_left(n4)
n2.add_right (n5)
n3.add_left(n6)
n3.add_right(n7)

print ("Preorder traversal result:", end=" ")
root.preorder_traversal()
```
