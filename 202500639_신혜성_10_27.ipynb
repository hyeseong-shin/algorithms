depth-first-paths

```python
class Graph:
  def __init__(self, V: int): # V: vertex개수
    self.V = V
    self.adj = [[] for _ in range(V)] # vertex마다 인접한 vertex를 저장하는 리스트 생성

  def add_edge(self, v: int, w: int): # edge추가하면 인접 리스트에 서로를 추가
    self.adj[v].append(w)
    self.adj[w].append(v)

  def adj_vertices(self, v: int): # 리스트에 들어있는 인접한 vertex들 반환
    return self.adj[v]
```

```python
class DepthFirstPaths:
  def __init__(self, G: Graph, s: int): # Graph 객체 입력받아 활용, s:시작점
    self.marked = [False] * G.V # 지나갔는지, vertex마다 생성
    self.edgeTo = [None] * G.V #어디서 왔는지
    self.s = s
    self.dfs(G, s)

  def dfs(self, G:Graph, v: int):
    """재귀적으로 DFS 수행"""
    self.marked[v] = True

    for a in G.adj_vertices(v):
      #방문하지 않은 인접한 vertex로 이동
      if self.marked[a] is False:
        self.edgeTo[a] = v
        self.dfs(G, a) #재귀호출(반복적으로 막힐 때까지)

  def path_to(self, v: int):
    """시작점 s로부터 v까지의 경로를 edgeTo로 복원"""
    if self.marked[v] is not True:
      return None #dfs에서 방문하지 못했으므로 경로가 없음

    path = []
    p = v #현재 vertex저장

    while p != self.s: #시작점에 도착할 때까지 이전 vertex로 역추적
      path.append(p)
      p = self.edgeTo[p]

    path.append(self.s) #시작점 추가
    path.reverse() #시작점부터 시작
    return path

if __name__ == "__main__":
  G = Graph(6)
  G.add_edge(0, 1)
  G.add_edge(0, 2)
  G.add_edge(0, 5)
  G.add_edge(1, 3)
  G.add_edge(3, 4)
  G.add_edge(4, 5)

  s = 0
  dfs_paths = DepthFirstPaths(G, s)

  for v in range(G.V):
    print(F"marked[{v}] = {dfs_paths.marked[v]}")

  #예시: 0에서 5까지 경로 출력
  print("path 0 -> 4:", dfs_paths.path_to(5))
```

connected components

```python
class CC:
  def __init__(self, G: Graph):
    self.marked = [False] * G.V
    self._id = [-1] * G.V
    self._count: int = 0

    #run DFS
    for v in range(G.V):
      if self.marked[v] is False:
        self.dfs(G, v)
        self._count += 1 #dfs끝나면 연결요소 하나 탐색한 것

  def count(self) -> int:
    """연결 요소의 개수"""
    return self._count

  def id(self, v:int) -> int:
    """정점 v가 속한 연결 요소 id (0..count-1)"""
    return self._id[v]

  def connected(self, v: int, w: int) -> bool:
    """정점v,w가 같은 연결 요소인지"""
    return self._id[v] == self._id[w]

  def dfs(self, G:Graph, v: int) -> None:
    self.marked[v] = True
    self._id[v] = self._count

    for a in G.adj_vertices(v):
      #방문하지 않은 인접한 vertex로 이동
      if self.marked[a] is False:
        self.dfs(G, a) #재귀호출(반복적으로 막힐 때까지)
```
