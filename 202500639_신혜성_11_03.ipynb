```python
class DiGraph: #방향 그래프
  def __init__(self, V: int):
    self.V = V
    self.adj = [[] for _ in range(V)]
  def add_edge(self, v: int, w: int):
    self.adj[v].append(w) #v->w

  def adj_vertices(self, v: int):
    return self.adj[v]
```

Multiple-source shortest paths

```python
from collections import deque

class BreadFirstMultiSourcePaths:
  def __init__(self, G, sources):
    self.marked = [False] * G.V
    self.edgeTo = [None] * G.V #해당 정점의 직전 정점
    self.disTo = [None] * G.V #가장 가까운 시작점까지의 거리
    self.sources = sources #여러 개의 시작 정점

    self.bfs(G, sources)

  def bfs(self, G, sources):
    """BFS from all sources at the same time"""
    q = deque()
    for s in sources:
      self.marked[s] = True
      self.disTo[s] = 0
      q.append(s)

    while q: #큐가 비어 있지 않은 동안
      v = q.popleft() #큐의 맨 앞 정점 꺼내 v에 저장

      for w in G.adj[v]: #v와 연결된 모든 인접 정점 w확인
        if not self.marked[w]: #방문하지 않은 경우(False)
          self.marked[w] = True #중복 방문 방지
          self.edgeTo[w] = v #최단 경로 복원
          self.disTo[w] = self.disTo[v] + 1 #최단 거리 계산
          q.append(w) #다음 탐색 대상으로 큐에 추가

  def has_path_to(self, v: int) -> bool:
    """Is v reachable from any source?"""
    return self.marked[v] #bfs에서 방문한 정점은 True로 변경됨

  def dist_to(self, v: int):
    """Distance from the nearest source to v"""
    return self.disTo[v]

  def path_to(self, v: int):
    """Shortest path from a source to v"""
    if not self.has_path_to(v):
      return None
    path = []
    x = v

    while x is not None: #시작점까지
      path.append(x)
      x = self.edgeTo[x] #직전 정점으로 이동

    path.reverse()
    return path
```

```python
#build the graph
G = DiGraph(7)
G.add_edge(0, 1)
G.add_edge(0, 2)
G.add_edge(1, 3)
G.add_edge(2, 3)
G.add_edge(3, 4)
G.add_edge(4, 5)
G.add_edge(2, 6)

bfs_paths = BreadFirstMultiSourcePaths(G, [0, 2])
for v in range(G.V):
  print(f"Vertex {v}: dist={bfs_paths.dist_to(v)}, path={bfs_paths.path_to(v)}")
```

Topological sort

```python
class TopologicalSort:
  def __init__(self, G):
    self.G = G
    self.marked = [False] * G.V
    self.order_list = [] # keeps the postorder

    # run DFS, build order_list
    for v in range(G.V):
      if not self.marked[v]:
        self.dfs(v)

    self.order_list.reverse()

  def dfs(self, v):
    self.marked[v] = True

    for w in self.G.adj[v]:
      if not self.marked[w]:
        self.dfs(w)

    self.order_list.append(v)


  def order(self):
    return self.order_list
```

```python
# a DAG to try it on
G = DiGraph(6)
G.add_edge(5, 2)
G.add_edge(5, 0)
G.add_edge(4, 0)
G.add_edge(4, 1)
G.add_edge(2, 3)
G.add_edge(3, 1)
topo = TopologicalSort(G)
print(topo.order())
```
