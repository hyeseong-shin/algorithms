Merge sort

```python
"""top-down merge sort방식
merge_sort(array): 재귀호출로 분할하고 정렬된 부분 리스트 병합
merge(left, right): 정렬된 두 리스트 원소 비교해 하나의 정렬된 리스트로 만들기"""

def merge_sort(array):
  #재귀호출로 원소 하나될 때까지 분할
  if len(array) <= 1:
    return array #현재 재귀 호출을 종료하고 호출한 위치로 값 반환

  mid = len(array) // 2
  left = merge_sort(array[:mid]) #분할한 왼쪽을 merge_sort로 다시 분할(재귀호출)
  right = merge_sort(array[mid:])

  return merge(left, right) #두 리스트를 병합한 결과를 상위 merge_sort 호출로 반환


def merge(left, right): #merge_sort의 보조 함수
  result = []
  i = 0
  j = 0
  #왼쪽, 오른쪽 리스트 원소 비교하며 작은 값을 새로운 리스트에 추가
  while i < len(left) and j < len(right):
    if left[i] <= right[j]:
      result.append(left[i])
      i += 1
    else:
      result.append(right[j])
      j += 1
  #남은 원소 추가(남은 한 리스트의 원소는 새 리스트의 원소보다 크고 정렬되어 있음)
  result.extend(left[i:])
  result.extend(right[j:])

  return result
```

```python
arr = [38, 27, 43, 3, 9, 82, 10]
print("정렬 전:", arr)
sorted_arr = merge_sort(arr)
print("정렬 후:", sorted_arr)
```

Quick sort

```python
"""in-place quick sort로 기존 배열의 원소 직접 교환하는 방식 이용
quick_sort: artition을 호출하고 Pivot의 왼쪽과 오른쪽에 재귀 호출
partition: 첫 번째 원소를 Pivot으로 선택하고,
i, j로 탐색·교환한 뒤 Pivot의 최종 인덱스 반환"""

def quick_sort(arr, low, high):
  if low >= high:
    return arr #재귀호출 종료

  p = partition(arr, low, high) #p에 partiton이 반환환 pivot인덱스

  quick_sort(arr, low, p-1) #기존 리스트에서 구역별로 정렬
  quick_sort(arr, p+1, high) #재귀호출

  return arr #위 두 재귀호출 종료되면 정렬된 리스트 반환

def partition(arr, low, high):
  v = low
  i = low + 1
  j = high

  while True:
    while i <= high and arr[i] < arr[v]:
      i += 1
    while j > low and arr[j] > arr[v]:
      j -= 1

    if i >= j:
      break
    arr[i], arr[j] = arr[j], arr[i]
    i += 1
    j -= 1

  arr[v], arr[j] = arr[j], arr[v]

  return j #pivot값의 위치
```

```python
arr = [38, 27, 43, 3, 9, 82, 10]
print("정렬 전:", arr)
quick_sort(arr, 0, len(arr) - 1)
print("정렬 후:", arr)
```
