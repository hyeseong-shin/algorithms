Bubble sort

```python
def bubble_sort(list):
  for i in range(len(list)-1):
    for j in range(len(list)-(i+1)):
      if list[j] > list[j+1]:
        list[j], list[j+1] = list[j+1], list[j] #값 교체

  return list
```

```python
my_list = [64, 34, 25, 12, 22, 11, 90]
print("Bubble 정렬 전:", my_list) #Bubble 정렬 전: [64, 34, 25, 12, 22, 11, 90]
sorted_list = bubble_sort(my_list)
print("Bubble 정렬 후:", sorted_list) #Bubble 정렬 후: [11, 12, 22, 25, 34, 64, 90]
```

Selection sort

```python
def selection_sort(list):
    for i in range(len(list)-1):
        min_idx= i #최소값 자체가 아니라 최소값의 인덱스 추적
        for j in range(i+1,len(list)):
            if list[j] < list[min_idx]:
                min_idx = j
        list[i], list[min_idx] = list[min_idx], list[i]

    return list
```

```python
my_list = [64, 34, 25, 12, 22, 11, 90]
print("Selection 정렬 전:", my_list) #Selection 정렬 전: [64, 34, 25, 12, 22, 11, 90]
sorted_list = selection_sort(my_list)
print("Selection 정렬 후:", sorted_list) #selection 정렬 후: [11, 12, 22, 25, 34, 64, 90]
```

Insertion sort

```python
def insertion_sort(list):
  for i in range(1, len(list)):
    m = list[i]
    j = i - 1 #오른쪽에서 왼쪽으로 shift
    while m < list[j] and j >= 0: #m < list[j]조건을 위해 for문x while문 사용
      list[j + 1] = list[j] #shift
      j -= 1
    list[j+1] = m

  return list
```

```python
my_list = [64, 34, 25, 12, 22, 11, 90]
print("Insertion 정렬 전:", my_list) #Insertion 정렬 전: [64, 34, 25, 12, 22, 11, 90]
sorted_list = insertion_sort(my_list)
print("Insertion 정렬 후:", sorted_list) #Insertion 정렬 후: [11, 12, 22, 25, 34, 64, 90]
```
